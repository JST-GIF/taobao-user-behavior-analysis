```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.sans-serif']=['SimHei']
rcParams['axes.unicode_minus']=False
# 从查询1结果导出的CSV读取
df = pd.read_csv('../data/retention_cohort.csv', parse_dates=['首次活跃日'])

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(df['首次活跃日'], df['rate_1d'], marker='o', label='次日留存率')
ax.plot(df['首次活跃日'], df['rate_3d'], marker='s', label='3日留存率')
ax.set_title('每日 cohort 留存率趋势（2017.11.25-12.03）')
ax.set_ylabel('留存率')
ax.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('retention_curve.png', dpi=150)
```
