```python
import pandas as pd
import random
#由于数据太大，因此100万为一轮进行读取
sample_df=pd.read_csv(
    '../data/UserBehavior.csv',
    names=['user_id', 'item_id', 'category_id', 'behavior_type', 'timestamp'],
    chunksize=1000000,
    iterator=True
)
#查看数据源到底有多少用户
target_users=set()                                                 
for chunk in sample_df:
    target_users.update(chunk['user_id'].unique())
print(f"在抽取之前一共有{len(target_users)}个用户")
#在里面抽取1万个用户来作为样本
sample_users=set(random.sample(sorted(target_users),10000))
print(f"已随机抽取{len(sample_users)}个用户作为样本")
#这时还要从新进行一轮读取，找出样本数据的所有记录
sample_df=pd.read_csv(
    '../data/UserBehavior.csv',
    names=['user_id', 'item_id', 'category_id', 'behavior_type', 'timestamp'],
    chunksize=1000000,
    iterator=True
)
#以100万为一轮的进行查询
final_df = pd.DataFrame()
chunks = []
for chunk in sample_df:
    filtered = chunk[chunk['user_id'].isin(sample_users)]
    chunks.append(filtered)
final_df=pd.concat(chunks,ignore_index=True)
#再次检验一下是否是1万的用户
print(f"样本用户的数量:{final_df['user_id'].nunique()}")
#查询一下一共有多少条记录
print(f"样本一共有:{len(final_df)}条记录")
#将抽取的数据提取出来
final_df.to_csv('../data/UserBehavior_sample.csv')
```
